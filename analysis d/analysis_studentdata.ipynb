```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Student": ["A","B","C","D","E","F","G","H"],
    "Age": [20,21,19,22,20,23,21,22],
    "Marks": [70,85,60,90,88,75,95,80],
    "Hours": [2,4,1,6,5,3,7,4]
})
```

```python
fig, ax = plt.subplots(4, 2, figsize=(20, 16))
ax[0, 0].scatter(df["Age"], df["Marks"])
ax[0,1].pie(df["Marks"], labels=df["Student"], autopct='%1.1f%%')
ax[1, 0].bar(df["Student"], df["Marks"])
ax[1,1].hist(df["Marks"], bins=5, edgecolor='black')
ax[2, 0].boxplot(df["Marks"])
ax[2,1].plot(df["Hours"], df["Marks"], marker='o')
ax[3, 0].barh(df["Student"], df["Marks"])
```

```python
fig1, ax1 = plt.subplots(4, 2, figsize=(20, 16))
sns.scatterplot(data=df, x="Age", y="Marks", ax=ax1[0, 0])
sns.barplot(data=df, x="Student", y="Marks", ax=ax1[1, 0])
sns.histplot(data=df, x="Marks", bins=5, kde=True, ax=ax1[1, 1])
sns.boxplot(data=df, x="Marks", ax=ax1[2, 0])
sns.lineplot(data=df, x="Hours", y="Marks", marker='o', ax=ax1[2, 1])
sns.barplot(data=df, x="Student", y="Marks", ax=ax1[3, 0])
sns.violinplot(data=df, y="Marks", ax=ax1[3, 1])

```

```python
sns.pairplot(df)

plt.show()
```

```python
sns.regplot(
    data=df,
    x="Hours",
    y="Marks"
)

plt.show()

plt.show()
```

```python
sns.ecdfplot(df["Marks"])

plt.show()
```

```python
sns.jointplot(   #using jointplot to visualize the relationship between Hours and Marks
    data=df,
    x="Hours",
    y="Marks",
    kind="scatter"
)
```

```python
sns.swarmplot(y=df["Marks"])

plt.show()
```

```python
sns.stripplot(y=df["Marks"])

plt.show()
```

```python
sns.kdeplot(df["Marks"], fill=True)

plt.show()
```
