```python
df['TIME OCC_str']=df['TIME OCC'].astype(str).str.zfill(4) #zfill is used to fill the missing values with zeros to make it a 4 digit number
```

```python
df['hour'] = df['TIME OCC_str'].str[:2].astype(int) #extracting the hour from the time column
```

```python
df['TIME_FORMATTED']=pd.to_datetime(df['TIME OCC_str'], format='%H%M').dt.time #converting the time column to a 12 hour format
```

```python
df
```

```python
df['Date Rptd']=pd.to_datetime(df['Date Rptd'],format='%m/%d/%Y %I:%M:%S %p')
df['DATE OCC']=pd.to_datetime(df['DATE OCC'],format='%m/%d/%Y %I:%M:%S %p')
```

```python
df['year']=df['DATE OCC'].dt.year
```

```python
df['month']=df['DATE OCC'].dt.month_name()
df['day']=df['DATE OCC'].dt.day_name()
```

```python
df
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
plt.subplots_adjust(wspace=0.3, hspace=0.3)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(18, 14))
plt.subplots_adjust(wspace=0.45, hspace=0.35)
sns.countplot(data=df, x='hour', ax=axes[0, 0],hue='hour', palette='viridis')

axes[0, 0].set_title('Crimes Count by Hour of the Day', fontsize=14)
axes[0, 0].set_xlabel('hour of the day')
axes[0, 0].set_ylabel('crimes count')
top_areas = df['AREA NAME'].value_counts().head(10)
sns.barplot(x=top_areas.values,y=top_areas.index, ax=axes[0, 1],hue=top_areas.index, palette='Blues_r')
axes[0, 1].set_title('Top 10 Crime Areas', fontsize=14)
axes[0, 1].set_xlabel('crimes count')
axes[0, 1].set_ylabel('area name')
sns.countplot(data=df, x='day', ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Crimes Count by Day of the Week', fontsize=14)
axes[1, 0].set_xlabel('day of the week')
top_crimes = df['Crm Cd Desc'].value_counts().head(10)
sns.barplot(x=top_crimes.values,y=top_crimes.index, ax=axes[1, 1],hue=top_crimes.index, palette='Set3')
axes[1, 1].set_title('Top 10 Crimes', fontsize=14)
axes[1, 1].set_xlabel('crimes count')
axes[1, 1].set_ylabel('crime description')
```

```python
fig2,axes2 = plt.subplots(2, 2, figsize=(16, 6))
plt.subplots_adjust(wspace=0.3, hspace=0.3)
sns.histplot(data=df,x='LOCATION',ax=axes2[0,0],bins=30)

```
